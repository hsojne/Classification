# SVM
: How to find the largest possible margin

- margin : distance to boundary.
- support vector : the data that makes up the margin

> $ w^{T}x + b = 0 $ , w and b are values ​​found through learning

> $ |w^{T}x + b| = 1 $, $ y_{i}(w^{T}x_{i} + b) \geq 1 $   $ Margin = \frac{2}{||w||} $



### Algorithm

max a, b $ \frac{2}{||w||} $

subject $ y_{i}(w^{T}x_{i} + b) \geq   1 $

for $ 1 \leq i \leq $ n

So, min w $ \frac{1}{2}||w||^{2} $


To solve, need Lagrange multiplier method and dual problems

- Hard Margin
  - All data is nicely divided

  - Separable as a completely linear model without allowing any misclassification


- Soft Margin
  - Data is congested

  - Allow for some degree of misclassification

  - Penalty granted
  
  > min$_{w, b, \zeta} \frac{1}{2}||w||^{2} + C \sum_{i=1}^{n}{\zeta_{i}} $

  > subject $ y_{i}(w^{T}x_{i} + b) \geq   1 - \zeta_{i} $ and $ \zeta_{i} \geq 0 $ $ for 1 \lt i \leq n $

    - slack variable : Assign slack variables to each data point.Provides a numerical representation of how much a data point violates the margin.

      - Data that does not violate margins : $ \zeta_{i} = 0 $
      
      - Data that does violate margins :
        - From margin boundaries to decision boundaries $ 0 \lt \zeta_{i} \leq 1$
        - After decision boundarte : $ 1 \lt \zeta_{i} $
